# MAP VIEW

Curret map view

```python
import ee
import geemap
from dotenv import load_dotenv
import os

# Load environment variables
load_dotenv()
PROJECT_ID = os.getenv("earth_id")

# Initialize Earth Engine
try:
    ee.Initialize(project=PROJECT_ID)
except Exception:
    ee.Authenticate()
    ee.Initialize(project=PROJECT_ID)

# India boundary
india = (
    ee.FeatureCollection("FAO/GAUL/2015/level0")
    .filter(ee.Filter.eq("ADM0_NAME", "India"))
)

# Select year (CHANGE YEAR HERE)
YEAR = 2019

dataset = (
    ee.Image(f"COPERNICUS/Landcover/100m/Proba-V-C3/Global/{YEAR}")
    .select("discrete_classification")
    .clip(india)
)

# Visualization parameters
vis_params = {
    "min": 10,
    "max": 100,
    "palette": [
        "006400",  # Forest
        "ffbb22",  # Shrubland
        "ffff4c",  # Grassland
        "f096ff",  # Cropland
        "fa0000",  # Urban
        "b4b4b4",  # Bare
        "f0f0f0",  # Snow/Ice
        "0064c8",  # Water
    ],
}

# Map
m = geemap.Map(center=[22.5, 78.9], zoom=4)
m.addLayer(dataset, vis_params, f"India Land Cover {YEAR}")
m.addLayer(india, {}, "India Boundary")
m
```
